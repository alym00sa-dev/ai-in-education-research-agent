## AI/Technology Relevance Analysis

This section uses Claude AI to determine which studies are relevant to AI/technology-enhanced education research. Each study is analyzed based on its intervention name and citation to determine if it involves AI, machine learning, educational technology, or data-driven personalization.

```python
import os
from anthropic import Anthropic
import json
import time

# Load API key from parent directory .env file
import sys
sys.path.append('../..')
from dotenv import load_dotenv
load_dotenv('../../research_assistant/.env')

client = Anthropic(api_key=os.getenv('ANTHROPIC_API_KEY'))

# Get unique studies
unique_studies = df.groupby('s_StudyID').agg({
    's_Citation': 'first',
    's_Intervention_Name': 'first',
    's_Study_Rating': 'first'
}).reset_index()

print(f"Analyzing {len(unique_studies):,} unique studies for technology relevance...")
print("This may take a few minutes...\n")

# Function to check relevance using Claude
def check_relevance(intervention_name, citation):
    """Use Claude to determine if a study involves ANY educational technology."""
    
    prompt = f"""You are analyzing educational research studies to determine if they involve ANY form of educational technology.

Intervention Name: {intervention_name}

Citation: {citation[:500]}

Determine if this study is relevant to technology-enhanced education. Consider relevant if it uses:
- Computer software, apps, or digital platforms of ANY kind
- Online learning systems or websites
- Educational games, simulations, or multimedia
- Video-based instruction or multimedia content
- Learning management systems
- Intelligent tutoring systems or adaptive learning
- Data analytics or dashboards for education
- Virtual reality, AR, or simulations
- Mobile apps for learning
- ANY computer-based or digital intervention
- Educational technology tools of any kind

NOT relevant (only non-tech interventions):
- Traditional curricula without technology component
- Face-to-face teaching practices only
- Physical manipulatives or printed materials only
- Policy changes without technology
- Teacher training without technology focus
- Purely administrative or organizational changes

Return ONLY a JSON object with this exact format:
{{
  "is_relevant": true/false,
  "justification": "Brief 1-sentence explanation of what technology is used OR why it's not tech-related"
}}"""

    try:
        response = client.messages.create(
            model="claude-3-5-haiku-20241022",
            max_tokens=200,
            messages=[{"role": "user", "content": prompt}]
        )
        
        result_text = response.content[0].text.strip()
        # Parse JSON response
        result = json.loads(result_text)
        return result['is_relevant'], result['justification']
    except Exception as e:
        print(f"Error analyzing study: {e}")
        return None, f"Error: {str(e)}"

# Analyze each study (with rate limiting)
results = []
for idx, row in unique_studies.iterrows():
    if idx % 50 == 0:
        print(f"Progress: {idx}/{len(unique_studies)} ({idx/len(unique_studies)*100:.1f}%)")
    
    is_relevant, justification = check_relevance(
        row['s_Intervention_Name'], 
        row['s_Citation']
    )
    
    results.append({
        's_StudyID': row['s_StudyID'],
        'is_tech_relevant': is_relevant,
        'relevance_justification': justification
    })
    
    # Rate limiting - small delay between calls
    time.sleep(0.1)

print(f"\nAnalysis complete! Analyzed {len(results)} studies.")

# Convert to DataFrame and merge
relevance_df = pd.DataFrame(results)
df_with_relevance = df.merge(relevance_df, on='s_StudyID', how='left')

# Update original dataframe
df = df_with_relevance

print("\nRelevance columns added to dataframe: 'is_tech_relevant', 'relevance_justification'")
```

```python
# Summary of technology relevance analysis

print("="*80)
print("TECHNOLOGY RELEVANCE SUMMARY")
print("="*80)

# Get unique studies with relevance
unique_with_relevance = df.groupby('s_StudyID').agg({
    'is_tech_relevant': 'first',
    'relevance_justification': 'first',
    's_Intervention_Name': 'first',
    's_Citation': 'first'
    

# Count relevant vs not relevant
relevant_count = unique_with_relevance['is_tech_relevant'].sum()
not_relevant_count = (~unique_with_relevance['is_tech_relevant']).sum()
total = len(unique_with_relevance)

print(f"\nTotal studies analyzed: {total:,}")
print(f"Technology-relevant studies: {relevant_count:,} ({relevant_count/total*100:.1f}%)")
print(f"Not technology-relevant: {not_relevant_count:,} ({not_relevant_count/total*100:.1f}%)")

# Analyze reasons for NOT being relevant
print("\n" + "="*80)
print("MAIN REASONS FOR NON-RELEVANCE")
print("="*80)

not_relevant_studies = unique_with_relevance[~unique_with_relevance['is_tech_relevant']]
print(f"\nSample of non-relevant studies (showing first 10 of {len(not_relevant_studies)}):")
print("-"*80)
for idx, row in not_relevant_studies.head(10).iterrows():
    print(f"\nIntervention: {row['s_Intervention_Name']}")
    print(f"Reason: {row['relevance_justification']}")

# Show some examples of relevant studies
print("\n" + "="*80)
print("SAMPLE OF TECHNOLOGY-RELEVANT STUDIES")
print("="*80)

relevant_studies = unique_with_relevance[unique_with_relevance['is_tech_relevant']]
print(f"\nShowing 10 examples of {len(relevant_studies)} relevant studies:")
print("-"*80)
for idx, row in relevant_studies.head(10).iterrows():
    print(f"\nIntervention: {row['s_Intervention_Name']}")
    print(f"Technology: {row['relevance_justification']}")
```

## Mapping to Four Implementation Objectives

For technology-relevant studies, classify each intervention into one of our four broadened implementation objectives:
1. **Adaptive Instruction & Tutoring Systems** - Real-time feedback, personalized instruction, intelligent tutoring
2. **Personalized Learning & Advising Systems** - Student advising, course recommendations, personalized pathways
3. **Data-Driven Decision Support** - Analytics dashboards, predictive models, administrative decision tools
4. **Learning Pathways & Mobility Support** - Career navigation, skill assessment, program/credential selection

```python
# Map technology-relevant interventions to four implementation objectives

# Get unique tech-relevant studies
tech_relevant_studies = df[df['is_tech_relevant'] == True].groupby('s_StudyID').agg({
    's_Intervention_Name': 'first',
    's_Citation': 'first',
    'relevance_justification': 'first'
}).reset_index()

print(f"Mapping {len(tech_relevant_studies):,} technology-relevant studies to implementation objectives...")
print("This may take a few minutes...\n")

# Four implementation objectives
IMPLEMENTATION_OBJECTIVES = [
    "Adaptive Instruction & Tutoring Systems",
    "Personalized Learning & Advising Systems",
    "Data-Driven Decision Support",
    "Learning Pathways & Mobility Support"
]

def map_to_objective(intervention_name, citation, tech_description):
    """Use Claude to map intervention to one of four implementation objectives."""
    
    prompt = f"""You are classifying educational technology interventions into ONE of four categories.

Intervention Name: {intervention_name}

Citation: {citation[:500]}

Technology Description: {tech_description}

Classify this intervention into EXACTLY ONE of these categories:

1. "Adaptive Instruction & Tutoring Systems"
   - Intelligent tutoring systems (ITS)
   - Adaptive learning platforms
   - Real-time feedback and assessment
   - Personalized instruction and content delivery
   - Automated grading and instructional adjustment
   - Computer-aided instruction with adaptive features
   
2. "Personalized Learning & Advising Systems"
   - Student advising and guidance systems
   - Course recommendation engines
   - College/career counseling tools
   - Personalized learning pathway planning
   - Student success coaching platforms
   - Academic planning and scheduling tools
   
3. "Data-Driven Decision Support"
   - Learning analytics dashboards
   - Predictive analytics for student outcomes
   - Early warning systems for at-risk students
   - Administrative decision-making tools
   - Data visualization for educators/administrators
   - Performance monitoring and reporting systems
   
4. "Learning Pathways & Mobility Support"
   - Career navigation and exploration tools
   - Skill assessment and competency mapping
   - Credential/program selection tools
   - Transfer pathway guidance
   - Job/career matching systems
   - Workforce readiness platforms

Choose the BEST FIT based on the primary function of the intervention.

Return ONLY a JSON object with this exact format:
{{
  "implementation_objective": "Exact name from list above",
  "reasoning": "Brief 1-sentence explanation of why this category fits best"
}}"""

    try:
        response = client.messages.create(
            model="claude-3-5-haiku-20241022",
            max_tokens=200,
            messages=[{"role": "user", "content": prompt}]
        )
        
        result_text = response.content[0].text.strip()
        result = json.loads(result_text)
        return result['implementation_objective'], result['reasoning']
    except Exception as e:
        print(f"Error mapping study: {e}")
        return "Unknown", f"Error: {str(e)}"

# Map each tech-relevant study
mapping_results = []
for idx, row in tech_relevant_studies.iterrows():
    if idx % 50 == 0:
        print(f"Progress: {idx}/{len(tech_relevant_studies)} ({idx/len(tech_relevant_studies)*100:.1f}%)")
    
    objective, reasoning = map_to_objective(
        row['s_Intervention_Name'],
        row['s_Citation'],
        row['relevance_justification']
    )
    
    mapping_results.append({
        's_StudyID': row['s_StudyID'],
        'implementation_objective': objective,
        'objective_reasoning': reasoning
    })
    
    # Rate limiting
    time.sleep(0.1)

print(f"\nMapping complete! Mapped {len(mapping_results)} studies.")

# Convert to DataFrame and merge
mapping_df = pd.DataFrame(mapping_results)
df = df.merge(mapping_df, on='s_StudyID', how='left')

print("\nMapping columns added to dataframe: 'implementation_objective', 'objective_reasoning'")
```

```python
# Summary of mapping to four implementation objectives

print("="*80)
print("IMPLEMENTATION OBJECTIVE DISTRIBUTION")
print("="*80)

# Get unique studies with mappings
mapped_studies = df[df['implementation_objective'].notna()].groupby('s_StudyID').agg({
    'implementation_objective': 'first',
    'objective_reasoning': 'first',
    's_Intervention_Name': 'first'
}).reset_index()

# Count by objective
objective_counts = mapped_studies['implementation_objective'].value_counts()

print(f"\nTotal technology-relevant studies mapped: {len(mapped_studies):,}\n")
print("Distribution across four implementation objectives:")
print("-"*80)
for objective, count in objective_counts.items():
    percentage = (count / len(mapped_studies)) * 100
    print(f"{objective:50} {count:5,} ({percentage:5.1f}%)")

# Show examples for each objective
print("\n" + "="*80)
print("SAMPLE INTERVENTIONS BY OBJECTIVE")
print("="*80)

for objective in IMPLEMENTATION_OBJECTIVES:
    obj_studies = mapped_studies[mapped_studies['implementation_objective'] == objective]
    if len(obj_studies) > 0:
        print(f"\n{objective} ({len(obj_studies)} total)")
        print("-"*80)
        for idx, row in obj_studies.head(5).iterrows():
            print(f"\n  • {row['s_Intervention_Name']}")
            print(f"    Reasoning: {row['objective_reasoning']}")
    else:
        print(f"\n{objective} (0 studies)")
        print("-"*80)
        print("  No studies mapped to this objective")
```
